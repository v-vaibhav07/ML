```
### Data Dictionary

The matches dataset contains information about each IPL match.

id: Unique ID of each match.

season: Season in which the match was played.

city: City where the match was played.

date: Date of the match.

match_type: Type of the match.

player_of_match: Player who received the Player of the Match award.

venue: Place where the match was played.

team1: First team of the match.

team2: Second team of the match.

toss_winner: Team that won the toss.

toss_decision: Decision taken after winning the toss.

winner: Team that won the match.

result: Type of match result.

result_margin: Margin by which the match was won.

target_runs: Runs required by the chasing team.

target_overs: Overs available to chase the target.

super_over: Shows if there was a Super Over.

method: Special method used for the match result.

umpire1: First umpire.

umpire2: Second umpire.


### Deliveries Dataset

The deliveries dataset contains ball-by-ball information of the matches.

match_id: ID of the match.

inning: Innings number.

batting_team: Team that is batting.

bowling_team: Team that is bowling.

over: Over number.

ball: Ball number.

batter: Player facing the ball.

bowler: Player bowling the ball.

non_striker: Batter at the other end.

batsman_runs: Runs scored by the batter.

extra_runs: Extra runs given on the ball.

total_runs: Total runs scored on the ball.

extras_type: Type of extra runs.

is_wicket: Shows if a wicket happened.

player_dismissed: Name of the player who got out.

dismissal_kind: Type of dismissal.

fielder: Fielder involved in the dismissal.


### Connection Between the Two Datasets

The matches dataset has information about each match.

The deliveries dataset has ball-by-ball information.

Both datasets are connected using the match ID.

matches.id is connected with deliveries.match_id.

One match can have many delivery records.
```

```
def check_outliers(data, column):
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = data[
        (data[column] < lower) |
        (data[column] > upper)
    ]

    print(column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Number of outliers:", len(outliers))
    print()
```

```
check_outliers(match_stats, 'result_margin')
check_outliers(match_stats, 'total_runs')
check_outliers(match_stats, 'wickets')
check_outliers(match_stats, 'boundaries')
```

Task 3C — Aggregation and Merge.

```

# Task 3C - Aggregation and Merge

team_summary = deliveries.groupby(
    ['match_id', 'batting_team']
).agg(
    total_runs=('total_runs', 'sum'),
    total_balls=('ball', 'count'),
    total_wickets=('is_wicket', 'sum'),
    total_boundaries=('is_boundary', 'sum'),
    total_dot_balls=('is_dot_ball', 'sum')
).reset_index()

display(team_summary.head(10))
```

```
team_summary['run_rate'] = (
    team_summary['total_runs'] / team_summary['total_balls']
) * 6

team_summary['boundary_percentage'] = (
    team_summary['total_boundaries'] / team_summary['total_balls']
) * 100

team_summary['dot_ball_percentage'] = (
    team_summary['total_dot_balls'] / team_summary['total_balls']
) * 100

display(team_summary.head(10))
```

```
matches_working = matches.copy()

processed = matches_working.merge(
    team_summary,
    left_on='id',
    right_on='match_id',
    how='left'
)

print("Matches rows before merge:", len(matches_working))
print("Rows after merge:", len(processed))

display(processed.head(10))
```

```
print("Number of matches:", matches_working['id'].nunique())
print("Number of matches after merge:", processed['id'].nunique())

print("Missing match IDs after merge:",
      processed['match_id'].isnull().sum())

print("Rows per match:")
print(processed.groupby('id').size().value_counts().sort_index())
```

Task 3D — Team Performance Summary.

```
# Task 3D - Team Performance Summary

teams = sorted(
    set(matches['team1'].dropna()) |
    set(matches['team2'].dropna())
)

print("Number of teams:", len(teams))
print(teams)
```

```
team_data = []

for team in teams:

    matches_played = (
        (matches['team1'] == team) |
        (matches['team2'] == team)
    ).sum()

    wins = (matches['winner'] == team).sum()

    no_results = (
        (
            (matches['team1'] == team) |
            (matches['team2'] == team)
        ) &
        (matches['is_no_result'] == 1)
    ).sum()

    losses = matches_played - wins - no_results

    team_data.append([
        team,
        matches_played,
        wins,
        losses,
        no_results
    ])

team_performance = pd.DataFrame(
    team_data,
    columns=[
        'team',
        'matches_played',
        'wins',
        'losses',
        'no_results'
    ]
)

display(team_performance)
```

```
team_performance['win_percentage'] = (
    team_performance['wins'] /
    (team_performance['matches_played'] - team_performance['no_results'])
) * 100

team_performance['win_percentage'] = (
    team_performance['win_percentage'].round(2)
)

team_performance = team_performance.sort_values(
    'win_percentage',
    ascending=False
).reset_index(drop=True)

display(team_performance)
```

# Team Performance Summary

For each team, I calculated the total matches played, wins, losses, no-result matches, and win percentage.

Win percentage was calculated using only matches that produced a result:

Win Percentage = Wins / (Matches Played - No Results) × 100

Therefore, no-result matches were excluded from the denominator because neither team won the match.
